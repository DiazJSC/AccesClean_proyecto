## Extracción datos de carpeta compartida - Dataset Solicitudes Realizadas

```python
CARPETA_CREAR = "historico_solicitudes"
ORIGEN = Path(r"A:\Universidad\AccesClean\Solicitudes")
DESTINO = Path(raw_dir) / CARPETA_CREAR

if not DESTINO.exists():
    DESTINO.mkdir(parents=True)
    print(f"📁 Carpeta {CARPETA_CREAR} creada correctamente")
else:
    print(f"¡📁 La carpeta {CARPETA_CREAR} ya existe!")
```

```python
for archivo in ORIGEN.glob("*"):
    print("[1] Archivo(s) a copiar en carpeta - Nombre: "+archivo.name)
    
    if archivo.name.startswith("~$"):
        print("[2] Archivo temporal ignorado - Nombre: "+archivo.name)
        continue
        
    shutil.copy2(archivo, DESTINO)
    print("[3] Archivo(s) copiados exitosamente en destino - Nombre: "+archivo.name)
```

```python
archivos = [f for f in DESTINO.iterdir() if f.is_file()]

if not archivos:
    print("¡No hay archivos para procesar!")
else:
    print(f"¡✅ Se encontraron {len(archivos)} archivos para procesar!")
```
